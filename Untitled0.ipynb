````python
import json
import re

def extract_order(message):
    messages = [
        {
            "role": "system",
            "content": (
                "Extract restaurant orders from user messages. "
                "Return only a valid JSON object, without Markdown. "
                "Required keys: tableNumber, customerCount, items. "
                "Each item must have name and quantity. "
                "Do not invent missing values. "
                "Use null for unknown tableNumber or customerCount."
            )
        },
        {
            "role": "user",
            "content": message
        }
    ]

    result = model(
        text=messages,
        max_new_tokens=256
    )

    output = result[0]["generated_text"][-1]["content"].strip()

    # Remove Markdown code fences if the model adds them.
    output = re.sub(
        r"^```(?:json)?\s*|\s*```$",
        "",
        output
    ).strip()

    order = json.loads(output)

    assert isinstance(order["items"], list)

    for item in order["items"]:
        assert isinstance(item["name"], str)
        assert isinstance(item["quantity"], int)
        assert item["quantity"] > 0

    return order
````

````python

import json
import re
from typing import Optional

from pydantic import BaseModel, Field, ConfigDict, ValidationError


# ============================================
# 1. DEFINE ORDER SCHEMA
# ============================================

class OrderItem(BaseModel):
    model_config = ConfigDict(strict=True, extra="forbid")

    name: str = Field(min_length=1)
    quantity: int = Field(gt=0)


class RestaurantOrder(BaseModel):
    model_config = ConfigDict(strict=True, extra="forbid")

    tableNumber: Optional[int]
    customerCount: Optional[int] = Field(gt=0)
    items: list[OrderItem]


# ============================================
# 2. SYSTEM PROMPT
# ============================================

SYSTEM_PROMPT = """
You are a restaurant order extraction assistant.

Your job is to convert natural language restaurant
orders into structured JSON.

STRICT RULES:

1. Return ONLY valid JSON without Markdown.
2. Never invent missing information.
3. If the table number is not explicitly mentioned,
   return null.
4. If the customer count is not explicitly mentioned,
   return null.
5. Table numbers and quantities must be integers.
6. Never infer customer count from item quantities.
7. Extract only food and drinks requested by the user.
8. Never calculate prices.
9. Never add items that were not requested.
10. Preserve item quantities correctly.
11. If a field is unknown, use null rather than guessing.

REQUIRED OUTPUT FORMAT:

{
  "tableNumber": null,
  "customerCount": null,
  "items": [
    {
      "name": "example item",
      "quantity": 1
    }
  ]
}

EXAMPLE 1:

Input:
Table 2 needs three pizzas and two coffees.

Output:
{
  "tableNumber": 2,
  "customerCount": null,
  "items": [
    {"name": "pizza", "quantity": 3},
    {"name": "coffee", "quantity": 2}
  ]
}

EXAMPLE 2:

Input:
Add two burgers and one Coke.

Output:
{
  "tableNumber": null,
  "customerCount": null,
  "items": [
    {"name": "burger", "quantity": 2},
    {"name": "Coke", "quantity": 1}
  ]
}

EXAMPLE 3:

Input:
Four people at table 7 ordered four momos.

Output:
{
  "tableNumber": 7,
  "customerCount": 4,
  "items": [
    {"name": "momo", "quantity": 4}
  ]
}

EXAMPLE 4:

Input:
Table 5 has three customers. Two want chicken
momo and the other wants fried rice.

Output:
{
  "tableNumber": 5,
  "customerCount": 3,
  "items": [
    {"name": "chicken momo", "quantity": 2},
    {"name": "fried rice", "quantity": 1}
  ]
}
"""


# ============================================
# 3. EXTRACT ORDER FUNCTION
# ============================================

def extract_order(message: str) -> dict:

    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT
        },
        {
            "role": "user",
            "content": message
        }
    ]

    # Generate the model response
    response = model(
        text=messages,
        max_new_tokens=256,
        generate_kwargs={
            "do_sample": False
        }
    )

    # Extract generated text
    output = response[0]["generated_text"][-1]["content"]

    # Remove Markdown code blocks if present
    output = re.sub(
        r"^```(?:json)?\s*|\s*```$",
        "",
        output.strip()
    ).strip()

    # Parse JSON
    data = json.loads(output)

    # Validate against Pydantic schema
    order = RestaurantOrder.model_validate(data)

    return order.model_dump()


# ============================================
# 4. TEST CASES
# ============================================

tests = [
    "Table 2 needs three pizzas and two coffees.",

    "We have four people on table 7. "
    "Two want chicken momo and the others want fried rice.",

    "Add one Coke and two burgers to my order.",

    "Table 5 has 3 customers. "
    "They want 2 chicken momos, 1 burger and 3 Cokes.",

    "Give me two coffees and one sandwich.",

    "There are six customers at table 9. "
    "Three ordered chicken burgers and the remaining "
    "three ordered veg momo."
]


# ============================================
# 5. RUN TESTS
# ============================================

for message in tests:

    print("\n" + "=" * 60)
    print("INPUT:", message)

    try:
        result = extract_order(message)

        print("OUTPUT:")
        print(json.dumps(result, indent=2))

    except (ValidationError, ValueError, KeyError, TypeError) as e:
        print("VALIDATION FAILED:", e)

print("\nTesting completed.")
````

````python

import re
from decimal import Decimal


def normalize_name(name: str) -> str:
    name = name.lower().strip()
    name = re.sub(r"[^a-z0-9\s]", "", name)
    return " ".join(name.split())


def match_menu_item(item_name: str, menu: list) -> dict:
    requested = normalize_name(item_name)

    # Handle simple plural forms
    variants = {requested}

    if requested.endswith("s"):
        variants.add(requested[:-1])

    matches = []

    for menu_item in menu:
        menu_name = normalize_name(menu_item["name"])

        if menu_name in variants:
            matches.append(menu_item)

    if len(matches) == 1:
        item = matches[0]

        if not item["available"]:
            return {
                "status": "unavailable",
                "requestedName": item_name,
                "menuItemId": item["id"]
            }

        return {
            "status": "matched",
            "requestedName": item_name,
            "menuItemId": item["id"],
            "name": item["name"],
            "unitPrice": item["price"]
        }

    # Find possible options for an ambiguous name
    possible = [
        item
        for item in menu
        if requested in normalize_name(item["name"])
    ]

    if possible:
        return {
            "status": "ambiguous",
            "requestedName": item_name,
            "options": [
                {
                    "id": item["id"],
                    "name": item["name"],
                    "available": item["available"]
                }
                for item in possible
            ]
        }

    return {
        "status": "not_found",
        "requestedName": item_name
    }


def prepare_order_draft(message: str) -> dict:
    extracted = extract_order(message)

    matched_items = []
    issues = []
    subtotal = Decimal("0")

    for requested_item in extracted["items"]:
        match = match_menu_item(
            requested_item["name"],
            restaurant_menu
        )

        if match["status"] != "matched":
            issues.append({
                **match,
                "quantity": requested_item["quantity"]
            })
            continue

        quantity = requested_item["quantity"]
        unit_price = Decimal(str(match["unitPrice"]))
        line_total = unit_price * quantity

        matched_items.append({
            "menuItemId": match["menuItemId"],
            "name": match["name"],
            "quantity": quantity,
            "unitPrice": str(unit_price),
            "lineTotal": str(line_total)
        })

        subtotal += line_total

    return {
        "tableNumber": extracted["tableNumber"],
        "customerCount": extracted["customerCount"],
        "items": matched_items,
        "issues": issues,
        "subtotal": str(subtotal),
        "readyForConfirmation": (
            len(issues) == 0 and len(matched_items) > 0
        )
    }
````

````python

import json
import re
from decimal import Decimal
from pydantic import BaseModel, ConfigDict, Field


# Our temporary in-memory conversation state
order_session = None


def get_subtotal(items):
    return sum(
        (
            Decimal(item["unitPrice"]) * item["quantity"]
            for item in items
        ),
        Decimal("0")
    )


def show_order(session):
    print("\n--- CURRENT ORDER ---")
    print("Table:", session["tableNumber"])
    print("Status:", session["status"])

    for item in session["items"]:
        print(f'{item["quantity"]} x {item["name"]}')

    print("Subtotal: Rs.", get_subtotal(session["items"]))


def next_question(session):
    if not session["pendingItems"]:
        session["status"] = "awaiting_confirmation"

        return (
            f'Order subtotal: Rs. {get_subtotal(session["items"])}. '
            "Would you like to confirm?"
        )

    issue = session["pendingItems"][0]
    session["status"] = "awaiting_clarification"

    if issue["status"] == "ambiguous":
        options = ", ".join(
            item["name"]
            for item in issue["options"]
            if item["available"]
        )

        return (
            f'You requested {issue["quantity"]} '
            f'{issue["requestedName"]}. '
            f'Which would you like? Options: {options}'
        )

    if issue["status"] == "unavailable":
        return (
            f'{issue["requestedName"]} is unavailable. '
            "Please ask staff to remove or replace it."
        )

    return (
        f'{issue["requestedName"]} was not found on the menu. '
        "Please ask staff to remove or replace it."
    )


def start_order(message):
    global order_session

    draft = prepare_order_draft(message)

    order_session = {
        "tableNumber": draft["tableNumber"],
        "customerCount": draft["customerCount"],
        "items": draft["items"].copy(),
        "pendingItems": draft["issues"].copy(),
        "status": "draft"
    }

    show_order(order_session)
    print("\nASSISTANT:", next_question(order_session))

    return order_session
````

````python

class ClarificationSelection(BaseModel):
    model_config = ConfigDict(strict=True, extra="forbid")

    menuItemId: int
    quantity: int = Field(gt=0)


class ClarificationResult(BaseModel):
    model_config = ConfigDict(strict=True, extra="forbid")

    selections: list[ClarificationSelection]


def understand_clarification(message, pending_item):
    options = [
        {
            "menuItemId": item["id"],
            "name": item["name"]
        }
        for item in pending_item["options"]
        if item["available"]
    ]

    prompt = f"""
You resolve ambiguous restaurant menu choices.

The customer previously requested:
{pending_item["quantity"]} x {pending_item["requestedName"]}

Available options:
{json.dumps(options)}

Customer reply:
{message}

RULES:
1. Return only valid JSON.
2. Use only menuItemId values from available options.
3. Never invent items, quantities, or IDs.
4. Interpret quantities from the customer's reply.
5. If the reply is unclear, return an empty selections list.
6. If quantities are missing or uncertain, return
   an empty selections list.
7. Do not guess.

Example:
Options:
[
  {{"menuItemId": 1, "name": "Chicken Burger"}},
  {{"menuItemId": 2, "name": "Veg Burger"}}
]

Reply: One chicken and one veg.

Output:
{{
  "selections": [
    {{"menuItemId": 1, "quantity": 1}},
    {{"menuItemId": 2, "quantity": 1}}
  ]
}}

Now resolve the actual customer reply.
"""

    response = model(
        text=[
            {
                "role": "system",
                "content": (
                    "You are a precise menu clarification "
                    "parser. Return only JSON."
                )
            },
            {
                "role": "user",
                "content": prompt
            }
        ],
        max_new_tokens=256,
        generate_kwargs={"do_sample": False}
    )

    output = response[0]["generated_text"][-1]["content"]

    output = re.sub(
        r"^```(?:json)?\s*|\s*```$",
        "",
        output.strip()
    ).strip()

    parsed = ClarificationResult.model_validate(
        json.loads(output)
    )

    return parsed.model_dump()
````

````python

def reply_to_order(message):
    global order_session

    if order_session is None:
        print("No active order. Call start_order() first.")
        return

    if order_session["status"] != "awaiting_clarification":
        print("Order is not awaiting clarification.")
        return

    pending = order_session["pendingItems"][0]

    if pending["status"] != "ambiguous":
        print("This issue cannot be resolved by selection yet.")
        print(next_question(order_session))
        return

    try:
        result = understand_clarification(message, pending)
        selections = result["selections"]

        if not selections:
            print("ASSISTANT: I couldn't understand your choices.")
            print(next_question(order_session))
            return

        # Restrict choices to the options we supplied
        allowed_ids = {
            item["id"]
            for item in pending["options"]
            if item["available"]
        }

        selected_ids = [
            item["menuItemId"]
            for item in selections
        ]

        if (
            len(selected_ids) != len(set(selected_ids))
            or not set(selected_ids).issubset(allowed_ids)
        ):
            print("Invalid or duplicate menu item selection.")
            return

        # The quantity cannot silently change
        total_quantity = sum(
            item["quantity"] for item in selections
        )

        if total_quantity != pending["quantity"]:
            print(
                "ASSISTANT: Please clarify all",
                pending["quantity"],
                "items."
            )
            return

        # Prepare changes before updating the session
        new_items = []

        for selection in selections:
            menu_item = next(
                (
                    item for item in restaurant_menu
                    if item["id"] == selection["menuItemId"]
                ),
                None
            )

            if menu_item is None or not menu_item["available"]:
                print("Selected item is unavailable.")
                return

            unit_price = Decimal(str(menu_item["price"]))
            quantity = selection["quantity"]

            new_items.append({
                "menuItemId": menu_item["id"],
                "name": menu_item["name"],
                "quantity": quantity,
                "unitPrice": str(unit_price),
                "lineTotal": str(unit_price * quantity)
            })

        # Apply the validated changes
        order_session["items"].extend(new_items)
        order_session["pendingItems"].pop(0)

        show_order(order_session)

        print("\nASSISTANT:", next_question(order_session))

    except Exception as error:
        print("Unable to process clarification:", error)
        print("Please try describing your choices again.")
````

````python

import re

def confirm_order(message):
    global order_session

    if order_session is None:
        return "No active order."

    if order_session["status"] != "awaiting_confirmation":
        return "Your order is not ready for confirmation."

    # Normalize message
    response = message.lower().strip()

    # Remove punctuation and normalize spaces
    response = re.sub(r"[^\w\s]", "", response)
    response = " ".join(response.split())

    confirmations = {
        "yes",
        "confirm",
        "confirm order",
        "yes confirm",
        "yes confirm my order",
        "place order",
        "please confirm my order",
        "go ahead",
        "yes please"
    }

    cancellations = {
        "no",
        "cancel",
        "cancel order",
        "no cancel",
        "dont place it",
        "please cancel my order"
    }

    if response in confirmations:
        order_session["status"] = "confirmed"

        return (
            f'Order confirmed for table {order_session["tableNumber"]}. '
            f'Subtotal: Rs. {get_subtotal(order_session["items"])}'
        )

    if response in cancellations:
        order_session["status"] = "cancelled"
        return "Order draft cancelled."

    return "Please reply 'confirm' or 'cancel'."
````

````python

def chat(message):
    global order_session

    print(f"\nUSER: {message}")

    # No existing order, or previous order is finished
    if (
        order_session is None
        or order_session["status"] in ["confirmed", "cancelled"]
    ):
        start_order(message)
        return

    # Waiting for a menu clarification
    if order_session["status"] == "awaiting_clarification":
        reply_to_order(message)
        return

    # Waiting for customer confirmation
    if order_session["status"] == "awaiting_confirmation":
        response = confirm_order(message)
        print(f"\nASSISTANT: {response}")
        return

    print("\nASSISTANT: Unable to process the current order.")
````

````python

from uuid import uuid4
from decimal import Decimal

sessions = {}


def create_session(restaurant_id, table_number):
    if restaurant_id not in restaurants:
        raise ValueError("Restaurant does not exist")

    if not isinstance(table_number, int) or table_number <= 0:
        raise ValueError("Invalid table number")

    session_id = str(uuid4())

    sessions[session_id] = {
        "sessionId": session_id,
        "restaurantId": restaurant_id,
        "tableNumber": table_number,
        "customerCount": None,
        "items": [],
        "pendingItems": [],
        "status": "new"
    }

    return session_id


def calculate_subtotal(session):
    return sum(
        (
            Decimal(item["unitPrice"]) * item["quantity"]
            for item in session["items"]
        ),
        Decimal("0")
    )


def get_session(session_id):
    if session_id not in sessions:
        raise ValueError("Invalid session")

    return sessions[session_id]


def show_session(session_id):
    session = get_session(session_id)

    print("\nRESTAURANT:", restaurants[
        session["restaurantId"]
    ]["name"])

    print("TABLE:", session["tableNumber"])
    print("STATUS:", session["status"])

    for item in session["items"]:
        print(
            f'{item["quantity"]} x {item["name"]} '
            f'@ Rs. {item["unitPrice"]}'
        )

    print("SUBTOTAL: Rs.", calculate_subtotal(session))
````

````python

def chat_multi(session_id, message):
    session = get_session(session_id)

    restaurant = restaurants[session["restaurantId"]]
    menu = restaurant["menu"]

    print(f"\nUSER: {message}")

    # CASE 1: New order
    if session["status"] == "new":
        extracted = extract_order(message)

        # Trusted session controls table identity
        session["customerCount"] = extracted["customerCount"]

        # Reject conflicting table references
        if (
            extracted["tableNumber"] is not None
            and extracted["tableNumber"] != session["tableNumber"]
        ):
            print("ASSISTANT: This session belongs to a different table.")
            return

        for requested in extracted["items"]:
            match = match_menu_item(requested["name"], menu)

            if match["status"] == "matched":
                quantity = requested["quantity"]
                price = Decimal(str(match["unitPrice"]))

                session["items"].append({
                    "menuItemId": match["menuItemId"],
                    "name": match["name"],
                    "quantity": quantity,
                    "unitPrice": str(price),
                    "lineTotal": str(price * quantity)
                })
            else:
                session["pendingItems"].append({
                    **match,
                    "quantity": requested["quantity"]
                })

    # CASE 2: Resolve ambiguous menu item
    elif session["status"] == "awaiting_clarification":
        pending = session["pendingItems"][0]

        if pending["status"] != "ambiguous":
            print(
                "ASSISTANT: This item is unavailable or "
                "not on the menu. Please contact staff."
            )
            return

        result = understand_clarification(message, pending)
        selections = result["selections"]

        valid_options = {
            item["id"]: item
            for item in pending["options"]
            if item["available"]
        }

        ids = [s["menuItemId"] for s in selections]

        if (
            not selections
            or len(ids) != len(set(ids))
            or not set(ids).issubset(valid_options)
            or sum(s["quantity"] for s in selections)
                != pending["quantity"]
        ):
            print("ASSISTANT: Please clarify your selections.")
            return

        # Validate against this restaurant's current menu
        new_items = []

        for selection in selections:
            item = next(
                (
                    x for x in menu
                    if x["id"] == selection["menuItemId"]
                    and x["available"]
                ),
                None
            )

            if item is None:
                print("ASSISTANT: A selected item is unavailable.")
                return

            price = Decimal(str(item["price"]))
            quantity = selection["quantity"]

            new_items.append({
                "menuItemId": item["id"],
                "name": item["name"],
                "quantity": quantity,
                "unitPrice": str(price),
                "lineTotal": str(price * quantity)
            })

        session["items"].extend(new_items)
        session["pendingItems"].pop(0)

    # CASE 3: Confirm or cancel
    elif session["status"] == "awaiting_confirmation":
        import re

        normalized = re.sub(
            r"[^\w\s]", "", message.lower()
        )
        normalized = " ".join(normalized.split())

        if normalized in {
            "yes", "confirm", "confirm order",
            "yes confirm my order", "place order"
        }:
            session["status"] = "confirmed"
            print("ASSISTANT: Order confirmed in memory.")
            show_session(session_id)
            return

        if normalized in {"no", "cancel", "cancel order"}:
            session["status"] = "cancelled"
            print("ASSISTANT: Order draft cancelled.")
            return

        print("ASSISTANT: Please reply confirm or cancel.")
        return

    else:
        print("ASSISTANT: Session is already closed.")
        return

    # Determine next conversation step
    if session["pendingItems"]:
        issue = session["pendingItems"][0]
        session["status"] = "awaiting_clarification"

        if issue["status"] == "ambiguous":
            options = ", ".join(
                x["name"]
                for x in issue["options"]
                if x["available"]
            )
            print(f"ASSISTANT: Which would you like? {options}")
        else:
            print(
                "ASSISTANT: Cannot complete order. "
                f'{issue["requestedName"]} is '
                f'{issue["status"]}.'
            )
    elif session["items"]:
        session["status"] = "awaiting_confirmation"
        print(
            "ASSISTANT: Subtotal Rs.",
            calculate_subtotal(session),
            "- Confirm your order?"
        )
    else:
        print("ASSISTANT: No valid items were found.")
````

````python

from decimal import Decimal


def add_order_items(session_id, message):
    session = get_session(session_id)

    # 1. Prevent modifications to closed orders
    if session["status"] in ("confirmed", "cancelled"):
        return {
            "status": "rejected",
            "message": "This order is already closed."
        }

    # 2. Resolve existing ambiguities first
    if session["pendingItems"]:
        return {
            "status": "rejected",
            "message": "Resolve pending menu choices first."
        }

    # 3. Get the correct restaurant menu
    restaurant = restaurants[session["restaurantId"]]
    menu = restaurant["menu"]

    # 4. Extract the order using Qwen
    extracted = extract_order(message)

    # 5. Validate table number
    if (
        extracted["tableNumber"] is not None
        and extracted["tableNumber"] != session["tableNumber"]
    ):
        return {
            "status": "rejected",
            "message": "Table number does not match this session."
        }

    new_items = []
    issues = []

    # 6. Match requested food against the menu
    for requested in extracted["items"]:
        match = match_menu_item(
            requested["name"],
            menu
        )

        if match["status"] != "matched":
            issues.append({
                **match,
                "quantity": requested["quantity"]
            })
            continue

        quantity = requested["quantity"]
        price = Decimal(str(match["unitPrice"]))

        new_items.append({
            "menuItemId": match["menuItemId"],
            "name": match["name"],
            "quantity": quantity,
            "unitPrice": str(price),
            "lineTotal": str(price * quantity)
        })

    # 7. Update the in-memory order draft
    session["items"].extend(new_items)
    session["pendingItems"].extend(issues)

    # 8. Determine the next state
    if session["pendingItems"]:
        session["status"] = "awaiting_clarification"

    elif session["items"]:
        session["status"] = "awaiting_confirmation"

    # 9. Return the updated draft
    return {
        "status": session["status"],
        "items": session["items"],
        "pendingItems": session["pendingItems"],
        "subtotal": str(calculate_subtotal(session)),
        "message": (
            "Please resolve menu choices."
            if session["pendingItems"]
            else "Please confirm your updated order."
        )
    }


print("add_order_items() loaded successfully.")
````

````python

from decimal import Decimal


def resolve_pending_clarification(session_id, message):
    session = get_session(session_id)

    if not session["pendingItems"]:
        return {
            "status": "error",
            "message": "No pending clarification exists."
        }

    pending = session["pendingItems"][0]

    if pending["status"] != "ambiguous":
        return {
            "status": "needs_manual_resolution",
            "message": (
                f'{pending["requestedName"]} cannot be '
                "resolved through menu selection."
            )
        }

    try:
        result = understand_clarification(message, pending)
        selections = result["selections"]

        allowed_ids = {
            option["id"]
            for option in pending["options"]
            if option["available"]
        }

        selected_ids = [
            item["menuItemId"]
            for item in selections
        ]

        # Validate selections before changing state
        if (
            not selections
            or len(selected_ids) != len(set(selected_ids))
            or not set(selected_ids).issubset(allowed_ids)
            or sum(s["quantity"] for s in selections)
                != pending["quantity"]
        ):
            return {
                "status": "needs_clarification",
                "message": "Please clarify all requested items."
            }

        restaurant = restaurants[session["restaurantId"]]
        menu = restaurant["menu"]

        new_items = []

        for selection in selections:
            menu_item = next(
                (
                    item for item in menu
                    if item["id"] == selection["menuItemId"]
                    and item["available"]
                ),
                None
            )

            if menu_item is None:
                return {
                    "status": "rejected",
                    "message": "Selected menu item is unavailable."
                }

            quantity = selection["quantity"]
            price = Decimal(str(menu_item["price"]))

            new_items.append({
                "menuItemId": menu_item["id"],
                "name": menu_item["name"],
                "quantity": quantity,
                "unitPrice": str(price),
                "lineTotal": str(price * quantity)
            })

        # Commit the resolved items to the draft
        session["items"].extend(new_items)
        session["pendingItems"].pop(0)

        if session["pendingItems"]:
            session["status"] = "awaiting_clarification"
        else:
            session["status"] = "awaiting_confirmation"

        return {
            "status": session["status"],
            "items": session["items"],
            "subtotal": str(calculate_subtotal(session)),
            "message": (
                "Please resolve the remaining menu choices."
                if session["pendingItems"]
                else "Please confirm your updated order."
            )
        }

    except Exception as error:
        return {
            "status": "error",
            "message": str(error)
        }
````

````python

import re


def confirm_order_tool(session_id, message):
    session = get_session(session_id)

    # Only confirm orders awaiting confirmation
    if session["status"] != "awaiting_confirmation":
        return {
            "status": "rejected",
            "message": "This order is not ready for confirmation."
        }

    # Never confirm an order with unresolved items
    if session["pendingItems"]:
        return {
            "status": "rejected",
            "message": "Resolve all pending items first."
        }

    if not session["items"]:
        return {
            "status": "rejected",
            "message": "Cannot confirm an empty order."
        }

    normalized = re.sub(
        r"[^\w\s]", "",
        message.lower()
    )

    normalized = " ".join(normalized.split())

    allowed_phrases = {
        "confirm",
        "confirm order",
        "confirm my order",
        "yes confirm",
        "yes confirm my order",
        "place order",
        "yes place order"
    }

    if normalized not in allowed_phrases:
        return {
            "status": "needs_confirmation",
            "message": "Please explicitly confirm your order."
        }

    # In-memory confirmation only
    session["status"] = "confirmed"

    return {
        "status": "confirmed",
        "restaurant": restaurants[
            session["restaurantId"]
        ]["name"],
        "tableNumber": session["tableNumber"],
        "items": session["items"],
        "subtotal": str(calculate_subtotal(session)),
        "message": "Order confirmed successfully."
    }
````

````python

def cancel_order_v6(session_id):
    session = get_session(session_id)

    if session["status"] == "cancelled":
        return {
            "status": "rejected",
            "message": "The order is already cancelled."
        }

    if session["status"] == "confirmed":
        return {
            "status": "requires_staff_approval",
            "message": (
                "This order is already confirmed. "
                "Please contact restaurant staff "
                "to request cancellation."
            )
        }

    session["status"] = "cancelled"

    # Retain items for audit/display purposes.
    # No refund or kitchen cancellation is performed.
    return {
        "status": "cancelled",
        "message": "Your order draft has been cancelled."
    }


def clear_order_v6(session_id):
    session = get_session(session_id)

    if session["status"] in ("confirmed", "cancelled"):
        return {
            "status": "rejected",
            "message": (
                "A closed order cannot be cleared."
            )
        }

    session["items"] = []
    session["pendingItems"] = []
    session["customerCount"] = None
    session["status"] = "new"

    return {
        "status": "cleared",
        "tableNumber": session["tableNumber"],
        "subtotal": "0",
        "message": (
            "All draft items removed. "
            "You can start a new order."
        )
    }
````

````python

restaurant_reviews = {
    "restaurant_A": {
        1: {"rating": 4.8, "reviewCount": 120},
        2: {"rating": 4.4, "reviewCount": 45},
        3: {"rating": 4.6, "reviewCount": 90},
        4: {"rating": 4.2, "reviewCount": 30},
        5: {"rating": 4.0, "reviewCount": 20},
    },
    "restaurant_B": {
        1: {"rating": 4.7, "reviewCount": 80},
        2: {"rating": 4.3, "reviewCount": 40},
        3: {"rating": 4.9, "reviewCount": 110},
        4: {"rating": 4.1, "reviewCount": 25},
    }
}


def recommend_dishes_v6(session_id, limit=3):
    session = get_session(session_id)
    restaurant_id = session["restaurantId"]

    menu = restaurants[restaurant_id]["menu"]
    reviews = restaurant_reviews.get(restaurant_id, {})

    ranked = []

    for item in menu:
        if not item["available"]:
            continue

        review = reviews.get(item["id"])

        if not review:
            continue

        # Simple weighted rating to reduce the
        # impact of items with very few reviews.
        count = review["reviewCount"]
        rating = review["rating"]
        prior_rating = 4.0
        prior_weight = 20

        score = (
            count * rating
            + prior_weight * prior_rating
        ) / (count + prior_weight)

        ranked.append({
            "menuItemId": item["id"],
            "name": item["name"],
            "price": item["price"],
            "rating": rating,
            "reviewCount": count,
            "score": round(score, 3)
        })

    ranked.sort(
        key=lambda item: item["score"],
        reverse=True
    )

    return {
        "status": "success",
        "basis": "mock_review_ranking",
        "restaurant": restaurants[restaurant_id]["name"],
        "recommendations": ranked[:limit]
    }
````

````python

def extract_order_v6(message, menu):
    available_names = [
        item["name"]
        for item in menu
        if item["available"]
    ]

    prompt = """
You extract restaurant orders in English,
Nepali (Devanagari), and Romanized Nepali.

Return ONLY valid JSON.

Rules:
- Never invent items or quantities.
- Use English canonical food names.
- Singular/plural and Nepali number words must
  be converted to integer quantities.
- Do not assume customer count from food quantity.
- Use null for missing table or customer counts.
- If an item type is ambiguous, keep the general
  requested name instead of guessing.
- If the request cannot be understood, return
  an empty items list.
- Use only items explicitly requested.
- Do not calculate prices.

Examples:

Input: Table 5 ma dui ota chicken momo ra euta Coke.
Output:
{"tableNumber":5,"customerCount":null,
"items":[
{"name":"Chicken Momo","quantity":2},
{"name":"Coke","quantity":1}
]}

Input: मलाई दुई प्लेट चिकेन मम र एक कोक चाहियो।
Output:
{"tableNumber":null,"customerCount":null,
"items":[
{"name":"Chicken Momo","quantity":2},
{"name":"Coke","quantity":1}
]}

Input: Dui ota burger ra euta Coke.
Output:
{"tableNumber":null,"customerCount":null,
"items":[
{"name":"burger","quantity":2},
{"name":"Coke","quantity":1}
]}

Input: Hello.
Output:
{"tableNumber":null,"customerCount":null,"items":[]}
"""

    messages = [
        {
            "role": "system",
            "content": prompt
        },
        {
            "role": "user",
            "content": (
                "Restaurant menu item names:\n"
                + json.dumps(available_names, ensure_ascii=False)
                + "\n\nCustomer message:\n"
                + message
            )
        }
    ]

    response = model(
        text=messages,
        max_new_tokens=256,
        generate_kwargs={"do_sample": False}
    )

    output = response[0]["generated_text"][-1]["content"]
    output = re.sub(
        r"^```(?:json)?\s*|\s*```$",
        "",
        output.strip()
    ).strip()

    order = RestaurantOrder.model_validate(
        json.loads(output)
    )

    return order.model_dump()
````

````python

from decimal import Decimal


def add_order_items_v6(session_id, message):
    session = get_session(session_id)

    if session["status"] in ("confirmed", "cancelled"):
        return {
            "status": "rejected",
            "message": "Cannot modify a closed order."
        }

    if session["pendingItems"]:
        return {
            "status": "needs_clarification",
            "message": "Resolve existing items first."
        }

    menu = restaurants[session["restaurantId"]]["menu"]

    try:
        extracted = extract_order_v6(message, menu)
    except Exception:
        return {
            "status": "needs_clarification",
            "message": "I couldn't understand the order. Please repeat it."
        }

    if (
        extracted["tableNumber"] is not None
        and extracted["tableNumber"] != session["tableNumber"]
    ):
        return {
            "status": "rejected",
            "message": "The table number does not match your session."
        }

    if not extracted["items"]:
        return {
            "status": "needs_clarification",
            "message": "Please tell me which food and quantity you want."
        }

    new_items = []
    issues = []

    for requested in extracted["items"]:
        match = match_menu_item(requested["name"], menu)

        if match["status"] != "matched":
            issues.append({
                **match,
                "quantity": requested["quantity"]
            })
            continue

        price = Decimal(str(match["unitPrice"]))
        quantity = requested["quantity"]

        new_items.append({
            "menuItemId": match["menuItemId"],
            "name": match["name"],
            "quantity": quantity,
            "unitPrice": str(price),
            "lineTotal": str(price * quantity)
        })

    session["items"].extend(new_items)
    session["pendingItems"].extend(issues)

    session["status"] = (
        "awaiting_clarification"
        if session["pendingItems"]
        else "awaiting_confirmation"
    )

    return {
        "status": session["status"],
        "items": session["items"],
        "pendingItems": session["pendingItems"],
        "subtotal": str(calculate_subtotal(session)),
        "message": (
            "Please clarify your menu choices."
            if issues
            else "Please review and confirm your order."
        )
    }
````

````python

def agent_chat_v6(session_id, message):
    session = get_session(session_id)

    try:
        action = classify_intent_v6(message)
    except Exception:
        return {
            "status": "needs_clarification",
            "message": "I couldn't understand. Please rephrase."
        }

    # Cancellation and clearing take priority
    # even during pending clarification.
    if action == "cancel_order":
        return cancel_order_v6(session_id)

    if action == "clear_order":
        return clear_order_v6(session_id)

    # Handle allowed read-only interruptions.
    if action in AGENT_TOOLS:
        return {
            "action": action,
            "result": AGENT_TOOLS[action](session_id)
        }

    if action == "recommend_dish":
        return recommend_dishes_v6(session_id)

    if action == "greeting":
        return {
            "status": "success",
            "message": (
                "Namaste! Welcome to "
                + restaurants[session["restaurantId"]]["name"]
                + ". How can I help?"
            )
        }

    if action == "ask_for_help":
        return {
            "status": "success",
            "message": (
                "I can show the menu, take orders, "
                "recommend dishes and explain your bill."
            )
        }

    # A reply during clarification is interpreted
    # against the pending menu choices.
    if session["status"] == "awaiting_clarification":
        return resolve_pending_clarification(
            session_id, message
        )

    if action == "add_order_items":
        return {
            "action": action,
            "result": add_order_items_v6(
                session_id, message
            )
        }

    if action == "confirm_order":
         return {
            "action": action,
           "result": confirm_order_v6(
                session_id, message
           )
        }

    if action in (
        "remove_order_items",
        "modify_order_items"
    ):
        return {
            "status": "not_implemented",
            "action": action,
            "message": (
                "Item removal and replacement are "
                "not implemented in V6 yet."
            )
        }

    return {
        "status": "unsupported",
        "message": (
            "I couldn't understand that request. "
            "Please try English or Nepali, "
            "or ask a staff member for assistance."
        )
    }
````

````python

import unicodedata


CONFIRM_PHRASES = {
    # English
    "confirm",
    "confirm order",
    "confirm my order",
    "yes confirm",
    "yes confirm my order",
    "place order",
    "place my order",

    # Romanized Nepali
    "order confirm gara",
    "order confirm gardinu",
    "mero order confirm gara",
    "order pakka gara",
    "order pakka gardinu",
    "order pakka garnuhos",
    "mero order pakka gara",
    "order pathaideu",

    # Devanagari Nepali
    "अर्डर पक्का गर",
    "अर्डर पक्का गर्नुहोस्",
    "मेरो अर्डर पक्का गर्नुहोस्",
    "हुन्छ अर्डर पक्का गर्नुहोस्",
    "अर्डर कन्फर्म गर",
    "अर्डर कन्फर्म गर्नुहोस्"
}


def normalize_confirmation(text):
    text = unicodedata.normalize("NFC", text).casefold()

    # Remove punctuation while keeping Nepali vowel marks
    text = "".join(
        char if not unicodedata.category(char).startswith(
            ("P", "S")
        ) else " "
        for char in text
    )

    return " ".join(text.split())


NORMALIZED_CONFIRM_PHRASES = {
    normalize_confirmation(phrase)
    for phrase in CONFIRM_PHRASES
}


def confirm_order_v6(session_id, message):
    session = get_session(session_id)

    if session["status"] != "awaiting_confirmation":
        return {
            "status": "rejected",
            "message": "This order is not ready for confirmation."
        }

    if session["pendingItems"] or not session["items"]:
        return {
            "status": "rejected",
            "message": "The order is incomplete."
        }

    normalized = normalize_confirmation(message)

    if normalized not in NORMALIZED_CONFIRM_PHRASES:
        return {
            "status": "needs_confirmation",
            "message": (
                "Please explicitly confirm your order. "
                "You can say 'Confirm my order' or "
                "'अर्डर पक्का गर्नुहोस्'."
            )
        }

    # Only update in-memory state
    session["status"] = "confirmed"

    return {
        "status": "confirmed",
        "restaurant": restaurants[
            session["restaurantId"]
        ]["name"],
        "tableNumber": session["tableNumber"],
        "items": session["items"],
        "subtotal": str(calculate_subtotal(session)),
        "message": "Order confirmed successfully in memory."
    }
````

````python

import json
import re
from pydantic import BaseModel
from typing import Literal


class ContextIntent(BaseModel):
    intent: Literal[
        "agreement",
        "acknowledgement",
        "explicit_confirmation",
        "cancel_order",
        "rejection",
        "clarification_request",
        "unclear"
    ]


CONTEXT_SYSTEM_PROMPT = """
You are a Nepali restaurant conversation interpreter.

You understand:
- English
- Nepali (Devanagari)
- Romanized Nepali
- Informal Nepali slang
- Mixed Nepali and English

Classify the meaning of the customer's reply
using the previous assistant message.

POSSIBLE INTENTS:

agreement:
Customer agrees with a question or proposal.
Examples: huss, hunxa, okay, lala.
Only when context supports agreement.

acknowledgement:
Customer acknowledges information without
necessarily agreeing to an action.

explicit_confirmation:
Customer explicitly instructs the assistant
to confirm or place their restaurant order.

cancel_order:
Customer explicitly requests cancellation.

rejection:
Customer says no or disagrees.

clarification_request:
Customer asks what something means or
requests an explanation.

unclear:
The meaning cannot be determined reliably.

IMPORTANT RULES:
- Always consider the previous assistant message.
- Do not classify everything as acknowledgement.
- "Confirm my order" is explicit_confirmation.
- "k vaneko?" is clarification_request.
- "haina" usually means rejection or disagreement.
- Do not invent unsupported intentions.
- Return JSON only.

EXAMPLES:

Assistant: Order confirm garne?
Customer: huss
Output: {"intent":"agreement"}

Assistant: Ma menu dekhaunchu hai.
Customer: huss
Output: {"intent":"acknowledgement"}

Assistant: Order confirm garne?
Customer: confirm my order
Output: {"intent":"explicit_confirmation"}

Assistant: Order confirm garne?
Customer: haina cancel gardeu
Output: {"intent":"cancel_order"}

Assistant: Order confirm garne?
Customer: k vaneko?
Output: {"intent":"clarification_request"}

Assistant: Order confirm garne?
Customer: haina
Output: {"intent":"rejection"}

Assistant: Order confirm garne?
Customer: ???
Output: {"intent":"unclear"}
"""


def classify_context_v2(state, assistant, customer):
    messages = [
        {
            "role": "system",
            "content": CONTEXT_SYSTEM_PROMPT
        },
        {
            "role": "user",
            "content": json.dumps({
                "state": state,
                "assistantMessage": assistant,
                "customerReply": customer
            }, ensure_ascii=False)
        }
    ]

    response = model(
        text=messages,
        max_new_tokens=100,
        generate_kwargs={"do_sample": False}
    )

    output = response[0]["generated_text"][-1]["content"]

    output = re.sub(
        r"^```(?:json)?\s*|\s*```$",
        "",
        output.strip()
    ).strip()

    return ContextIntent.model_validate_json(
        output
    ).intent
````

````python

import json
from collections import Counter
from pathlib import Path

dataset_path = Path(
    "/content/restaurant_ai/starter_examples.jsonl"
)

allowed_intents = {
    "agreement",
    "acknowledgement",
    "explicit_confirmation",
    "cancel_order",
    "rejection",
    "clarification_request",
    "unclear"
}

records = []
seen = {}
errors = []
intent_counts = Counter()

with open(dataset_path, "r", encoding="utf-8") as file:
    for line_number, line in enumerate(file, start=1):
        try:
            record = json.loads(line)

            messages = record["messages"]

            if len(messages) != 3:
                raise ValueError(
                    "Expected exactly three messages"
                )

            roles = [m["role"] for m in messages]

            if roles != ["system", "user", "assistant"]:
                raise ValueError("Invalid message roles")

            context = json.loads(messages[1]["content"])
            answer = json.loads(messages[2]["content"])

            intent = answer["intent"]

            if intent not in allowed_intents:
                raise ValueError(
                    f"Unknown intent: {intent}"
                )

            key = (
                context["state"],
                context["assistantMessage"].strip().lower(),
                context["customerReply"].strip().lower()
            )

            if key in seen:
                if seen[key] != intent:
                    raise ValueError(
                        "Conflicting labels for same example"
                    )
                raise ValueError("Duplicate example")

            seen[key] = intent
            intent_counts[intent] += 1
            records.append(record)

        except Exception as error:
            errors.append({
                "line": line_number,
                "error": str(error)
            })

print("Valid examples:", len(records))
print("Errors:", len(errors))

print("\nIntent distribution:")

for intent, count in sorted(intent_counts.items()):
    print(f"{intent}: {count}")

if errors:
    print("\nValidation errors:")
    print(json.dumps(errors, indent=2))
else:
    print("\nDataset validation passed.")
````

````python

import json
import shutil
from collections import Counter
from pathlib import Path

from pydantic import BaseModel, ConfigDict, Field
from typing import Literal


Intent = Literal[
    "greeting", "order_request", "order_change",
    "order_confirmation", "order_cancellation",
    "menu_question", "recommendation", "complaint",
    "clarification", "acknowledgement", "unknown"
]

ActionType = Literal[
    "search_menu", "add_items", "remove_items",
    "replace_items", "clear_draft",
    "request_confirmation", "request_cancellation",
    "get_order_summary", "get_recommendations",
    "get_table_info"
]


class FoodItem(BaseModel):
    model_config = ConfigDict(extra="forbid")

    name: str = Field(min_length=1)
    quantity: int = Field(strict=True, gt=0)
    modifiers: list[str] = []


class Action(BaseModel):
    model_config = ConfigDict(extra="forbid")

    type: ActionType
    items: list[FoodItem]


class Decision(BaseModel):
    model_config = ConfigDict(extra="forbid")

    schemaVersion: Literal["1.0"]
    intent: Intent
    actions: list[Action] = Field(max_length=8)
    reply: str | None
    needsClarification: bool


class Context(BaseModel):
    model_config = ConfigDict(extra="forbid")

    state: str
    lastAssistantMessage: str | None


class TrainingExample(BaseModel):
    model_config = ConfigDict(extra="forbid")

    id: str
    language: Literal[
        "english", "nepali",
        "romanized_nepali", "mixed"
    ]
    context: Context
    customerMessage: str = Field(min_length=1)
    expected: Decision


# Load Gemini's generated JSON
with open("batch01.json", encoding="utf-8") as f:
    raw_data = json.load(f)

assert isinstance(raw_data, list), "Expected a JSON array"

valid = []
errors = []
seen_ids = set()
seen_messages = set()

for index, raw in enumerate(raw_data):
    try:
        record = TrainingExample.model_validate(raw)

        if record.id in seen_ids:
            raise ValueError("Duplicate ID")

        key = (
            record.context.state,
            record.context.lastAssistantMessage,
            record.customerMessage.casefold().strip()
        )

        if key in seen_messages:
            raise ValueError("Duplicate conversation")

        seen_ids.add(record.id)
        seen_messages.add(key)
        valid.append(record.model_dump())

    except Exception as error:
        errors.append({
            "index": index,
            "error": str(error)
        })

print("Generated:", len(raw_data))
print("Structurally valid:", len(valid))
print("Errors:", len(errors))

print("\nLanguages:")
print(Counter(x["language"] for x in valid))

print("\nIntents:")
print(Counter(x["expected"]["intent"] for x in valid))

if errors:
    print("\nValidation errors:")
    print(json.dumps(errors, indent=2, ensure_ascii=False))

# Save valid records for human review.
# Do not treat these as approved training data yet.
destination = Path(
    "/content/drive/MyDrive/restaurant_ai/datasets/v1"
)
destination.mkdir(parents=True, exist_ok=True)

output = destination / "batch01_unreviewed.jsonl"

with open(output, "w", encoding="utf-8") as f:
    for record in valid:
        f.write(
            json.dumps(record, ensure_ascii=False) + "\n"
        )

print("\nSaved unreviewed examples:", output)
````
